## 教程
本教程演示 HMFF 包的具体使用方法.
1. 导入 HMFF 包, 初始化 formfactors 字典
2. 获取 **具体过程** 的 FormFactor 对象
3. 获取关于此过程的某一篇文章中的形状因子数据

## 1. 导入 HMFF 包, 初始化 formfactors 字典

```python
import HMFF
from HMFF.classes import FormFactor, Impl   # 导入类型是为了后面的类型注释
formfactors = HMFF.formfactors
```

## 2. 获取 **具体过程** 的 FormFactor 对象 (以 B->K 过程为例)

### 2.1 查看 formfactors 中的信息

```python
# formfactors 是一个懒加载的字典, 可以重新加载
# formfactors.reload()

# HMFF.formfactors 这个字典内包含所有已经实现的过程信息：{'过程名'：形状因子对象}
print(formfactors.keys())
```

### 2.2 获取 B->K 过程的形状因子

```python
# 此时 ff_bk 对象内包含着 B->K 过程形状因子的所有信息
ff_bk: FormFactor = HMFF.formfactors['B->K']

# 查看 B->K 强子矩阵元参数化成的全部形状因子
# TODO: print(ff_bk.form_factor_names)

# 查看 B->K 过程形状因子收录了多少篇文章中的结果
print(ff_bk.impl_names)
```

## 3. 获取 B->K 过程的某一篇文章中的形状因子数据

```python
# 获取 B->K 过程形状因子: LQCD-FLAG-2024 这篇文章中的数据
ff_bk_impl: Impl = ff_bk.get_impl('LCSR-2004')

# 查看文章的引用
print(f"参考文献: {ff_bk_impl.ref}")

# 查看这篇文章中包含的所有形状因子
print(f"包含形状因子: {ff_bk_impl.form_factor_names}")
```

```python
# 获取其中一个形状因子的中心值
func_fplus = ff_bk_impl.form_factor_function("f+")
func_fplus(qsq=20)
```

```python
# 获取 q^2 = 20 GeV^2 时的所有形状因子的中心值
print(rf"$q^2 = 20 GeV^2$ 时的形状因子: {ff_bk_impl.get_central_values(qsq=20)}")
```
